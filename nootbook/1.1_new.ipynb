```python
df=df.drop('Year',axis=1)
```

```python
df=df.dropna()
X=df.drop('Selling_Price',axis=1)
y=df['Selling_Price']
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=42)
```

```python
X_train.shape,X_test.shape
```

```python
print(type(X_train))
```

```python
cat_features=X.select_dtypes(include='object').columns
num_features=X.select_dtypes(exclude='object').columns
```

```python
from sklearn.preprocessing import StandardScaler,OneHotEncoder
from sklearn.compose import ColumnTransformer
numeric_transformation=StandardScaler()
oh_transformation=OneHotEncoder(drop='first',handle_unknown='ignore',sparse_output=False)
preprocesser=ColumnTransformer(
    transformers=[
        ("OneHotEncoder",oh_transformation,cat_features),
        ("StandardScaler",numeric_transformation,num_features)
    ]
)
```

```python
import pickle
from sklearn.pipeline import Pipeline
from xgboost import XGBRegressor

pipeline = Pipeline(steps=[
    ("preprocessor", preprocesser),
    ("model", XGBRegressor())
])

pipeline.fit(X_train, y_train)

pickle.dump(pipeline, open("model.pkl", "wb"))
```

```python
X_train=preprocesser.fit_transform(X_train)
X_test=preprocesser.transform(X_test)
```

```python
X1=pd.DataFrame(X_train)
X1
```

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.linear_model import Lasso,Ridge
from sklearn.neighbors import KNeighborsRegressor
from sklearn.ensemble import GradientBoostingRegressor,AdaBoostRegressor,RandomForestRegressor
from xgboost import XGBRegressor
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

```python
models={
    "Linear":LinearRegression(),
    "Lasso":Lasso(),
    "Ridge":Ridge(),
    "Decision Tree":DecisionTreeRegressor(),
    "KNN":KNeighborsRegressor(),
    "Gradient Boost":GradientBoostingRegressor(),
    "AdaBoost":AdaBoostRegressor(),
    "Random Forest":RandomForestRegressor(),
    "XGB":XGBRegressor()
}
```

```python
for i in range(len(list(models))):
    model=list(models.values())[i]
    model.fit(X_train,y_train)
    # y_train_pred=(model.predict_proba(X_train)[:,1]>=0.5).astype(int)
    # y_test_pred=(model.predict_proba(X_test)[:,1]>=0.5).astype(int)
    y_train_pred=model.predict(X_train)
    y_test_pred=model.predict(X_test)
    
    model_train_mae=mean_absolute_error(y_train,y_train_pred)
    model_train_mean_squared_error=mean_squared_error(y_train,y_train_pred)
    model_train_r2_score=r2_score(y_train,y_train_pred)
  
    
    
    model_test_mae=mean_absolute_error(y_test,y_test_pred)
    model_test_mean_squared_error=mean_squared_error(y_test,y_test_pred)
    model_test_r2_score=r2_score(y_test,y_test_pred)

    
    print(list(models.keys())[i])
    
    print('Model performnace for traning set')
    print("- mae:{:.4f}".format(model_train_mae))
    print("- mse:{:.4f}".format(model_train_mean_squared_error))
    print("- r2_score:{:.4f}".format(model_train_r2_score))

    
    print("------------------------------------- ")
    
    print('Model performnace for testing set')
    
    print("- mae:{:.4f}".format(model_test_mae))
    print("- mse:{:.4f}".format(model_test_mean_squared_error))
    print("- r2_score:{:.4f}".format(model_test_r2_score))


    print('='*35)
    print('\n')

    
```

```python
trained_models = {}

for name, model in models.items():
    model.fit(X_train, y_train)
    
    y_train_pred = model.predict(X_train)
    y_test_pred = model.predict(X_test)

    print(name)

    trained_models[name] = model   # store fitted model
```
